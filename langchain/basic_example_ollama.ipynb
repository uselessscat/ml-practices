```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, SystemMessage

llm = ChatOllama(
    model="gemma3:27b",
    temperature=0,
)

messages = [
    SystemMessage("Traduce los siguiente desde Español al Inglés, las explicaciones deben estar en español"),
    HumanMessage("Hola amigos"),
]

ai_msg = llm.invoke(messages)
print(ai_msg.content)
```

```python
# for streaming
for token in llm.stream(messages):
    print(token.content, end="")
```

```python
from langchain_core.prompts import ChatPromptTemplate

system_template = "Translate the following from Spanish into {language}"

prompt_template = ChatPromptTemplate.from_messages(
    [("system", system_template), ("user", "{text}")]
)

prompt = prompt_template.invoke({"language": "Italian", "text": "Helo friends!"})

print(prompt.to_messages())

response = llm.invoke(prompt)
print(response.content)
```
