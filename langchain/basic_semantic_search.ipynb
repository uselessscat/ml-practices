```python
from langchain_core.documents import Document

documents = [
    Document(page_content="Perro: Compañero fiel y protector, ideal para familias activas. Requiere paseos diarios, afecto y entrenamiento."),
    Document(page_content="Gato: Independiente pero afectuoso. Perfecto para espacios pequeños. Necesita higiene, alimentación y algo de juego."),
    Document(page_content="Pez: Silencioso y decorativo. Vive en acuarios bien cuidados. Ideal para personas con poco tiempo."),
]

documents
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="bge-m3")

vectors = embeddings.embed_documents(list(map(lambda d: d.page_content, documents)))

len(vectors), len(vectors[0]), vectors[0][:10]
```

```python
from langchain_core.vectorstores import InMemoryVectorStore

vector_store = InMemoryVectorStore(embeddings)

vector_store.add_documents(documents=documents)
```

```python
results = vector_store.similarity_search_with_score(
    "Es excelente compañero?"
)

results
```

```python
from langchain_core.runnables import chain

@chain
def retriever(query: str):
    return vector_store.similarity_search(query, k=1)


retriever.batch(
    [
        "¿Quien es excelente compañero?",
        "¿Quien es independiente?",
        "¿Quien es pequeño?",
    ],
)
```
